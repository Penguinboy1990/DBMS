3. Explain how the Scikit-Learn ColumnTransformer and Pipeline streamline the data preparation process and prevent data leakage across preprocessing and training steps.

>* The pipelines allow us to chain the transformers together. For example, we can do imputation and scaling into one single object. While the ColumnTransformer applies different steps to specific feature types all at the same time. For example, the StandardScaler on numeric data and OneHotEncoder on categorical data. This lets us eliminate some of the manual steps and also maintains the exact matrix alignment.

>* The pipeline prevents data leakage by computing the parameters purely on the training set during .fit() and then applying those exact metrics to test sets during the .transform() so the data is isolated.

4. How did comparing StandardScaler versus MinMaxScaler impact your model training, convergence, or final evaluation metrics?

>* When comparing the two scalers, MinMaxScaler performed slightly better overall on this particular dataset. But both scalers had similar results on the 80/20 split (0.8391 vs 0.8277 ROC-AUC). MinMaxScaler achieved the highest peak ROC-AUC score of 0.8427 on the 70/30 split and performed well on the 60/40 split.

5. How does configuring SGDClassifier(loss='log_loss') differ from standard linear classification models, and what advantages do probability outputs and ROC-AUC metrics provide when evaluating customer churn?

>* SGDClassifiers with loss='log_loss' changes the model to a Logistic Regression model which is optimized with Stochastic Gradient Descent. The SGD provides some advantages such as the incremental weight updates. Which are ideal for dealing with large datasets that cannot fit into memory.

>* Probabilities are great here because they allow us to set flexible thresholds. We could also take advantage of stratification with probabilities, e.g., high, medium, low risk churners.

>* Since customer datasets are often imbalanced, where the vast majority of customers stay (no churn). We could theoretically get high accuracy despite the fact that a model can predict "no churn" for everyone. The ROC-AUC lets us test the model's performance across each possible classification threshold, as opposed to a single arbitrary cutoff.
