1) Why is feature scaling essential prior to executing K-Means, DBSCAN, or HDBSCAN on the Wine dataset? Specifically, how would unscaled features distort distance calculations relative to other features? Provide specific examples.

Feature scaling is important because these algorithms rely on distance based metrics, to measure how close two data points are. If the data variables are unscaled, they would heavily distort the distance calculations due to outliers dramatically impacting calculations. The features with the largest magnitudes control the shape and location of the clusters, thus rendering other equally important features more or less important unjustifyably.

DBSCAN and HDBSCAN can classify certain wine samples as noise because they are density based clustering algorithms and can label some samples as noise rather than belonging to a specific cluster. K-Means is flawed because it assigns each sample the cluster which it is closest to, leaving no outliers or noise behind.

4. Contrast the operational mechanisms of standard DBSCAN and HDBSCAN. How does HDBSCAN's hierarchical, stability-based approach overcome DBSCAN's vulnerability to a single global eps parameter when clustering chemical datasets with varying local densities?

HDBSCAN overcomes DBSCANS limitations by creating a hierarchical representation of clusters, across varying density levels. Instead of using one global epsilon value, it evaluates cluster stability using different density thresholds. The wine clusters which persist across a wider range of density levels are considered more stable and are selected as meaningful clusters. Since this dataset has varying local densities, HDBSCAN can use this to its advantage when clustering the data.
